# 🐼 Pandas Fundamentals: Data Inspection & Selection

## What You Will Learn
This notebook introduces the essential pandas operations every data analyst needs. We will cover:

1. **Data Inspection** — understanding structure, types, and missing values
2. **Positional Selection** with `iloc[]`
3. **Label-Based Selection** with `loc[]`
4. **Boolean Filtering** — selecting rows by conditions
5. **Slicing** — extracting subsets of rows and columns

## 1. Import Pandas

Pandas is the go-to library for tabular data manipulation in Python. The standard alias is `pd`.

```python
import pandas as pd

# Display version for reproducibility
print(f"Running pandas {pd.__version__}")
```

## 2. Load the Dataset

`pd.read_csv()` reads a comma-separated values file into a DataFrame. Common parameters include:
- `sep` — delimiter (default: comma)
- `header` — row to use as column names (default: 0)
- `index_col` — column to use as row labels
- `parse_dates` — columns to parse as datetime
- `na_values` — strings to recognize as NaN

```python
# Load products data from CSV
products = pd.read_csv('products.csv')

# Create the DataFrame
df = pd.DataFrame(products)

# Quick look at the data
df
```

## 3. Structural Overview with `df.info()`

`info()` prints a concise summary of the DataFrame, including:
- Total number of entries (rows)
- Column names and non-null counts
- Data types of each column
- Memory usage

This is your **first diagnostic tool** when receiving a new dataset.

```python
# Get a comprehensive summary of the DataFrame
df.info()
```

## 4. Accessing Column Names with `df.columns`

The `columns` attribute returns an Index object containing all column labels. This is useful for:
- Verifying expected columns exist
- Iterating over columns programmatically
- Renaming columns via list assignment

```python
# Retrieve all column names
print("Column names:")
print(df.columns)

# Convert to a plain Python list for easier manipulation
column_list = df.columns.tolist()
print("\nAs a list:", column_list)

# Count columns
print(f"\nTotal columns: {len(df.columns)}")
```

## 5. Checking Data Types with `df.dtypes`

`dtypes` shows the data type of each column. Common pandas types:

| Type | Description | Example |
|------|-------------|---------|
| `int64` | 64-bit integer | `Stock`, `Warranty_Months` |
| `float64` | 64-bit float | `Price`, `Rating` |
| `object` | Text / mixed types | `Name`, `Brand`, `Color` |
| `bool` | Boolean True/False | — |
| `datetime64[ns]` | Date and time | — |

> ⚠️ **Watch out:** Columns with mixed types are often cast to `object`, which can slow operations.

```python
# Display the data type of each column
print("Data types:")
print(df.dtypes)

# Check if any column needs type conversion
print("\nColumns with object dtype (potential strings):")
print(df.dtypes[df.dtypes == 'object'])
```

## 6. Getting Dimensions with `df.shape`

`shape` returns a tuple `(rows, columns)` representing the DataFrame dimensions. This is a **property** (not a method), so no parentheses needed.

```python
# Get the shape tuple
rows, cols = df.shape

print(f"DataFrame shape: {df.shape}")
print(f"Rows: {rows}")
print(f"Columns: {cols}")

# Total cells
print(f"\nTotal data cells: {rows * cols}")
```

## 7. Previewing the Top Rows with `df.head()`

`head(n)` returns the first `n` rows. If `n` is omitted, it defaults to 5. Use this to quickly sanity-check your data after loading.

```python
# Display the first 5 rows (default)
print("First 5 rows:")
display(df.head())

# Explicitly request 5 rows
print("\nExplicit head(5):")
df.head(5)
```

## 8. Previewing the Bottom Rows with `df.tail()`

`tail(n)` returns the last `n` rows. This is especially useful for checking recent entries in time-series data or verifying that a file was read completely.

```python
# Display the last 6 rows
print("Last 6 rows:")
df.tail(6)
```

## 9. Selecting a Single Cell with `iloc[]`

`iloc` stands for **integer location**. It selects data by **position** (0-based indexing), regardless of what the index labels actually are.

**Syntax:** `df.iloc[row_index, column_index]`

In this example, `iloc[0, 1]` selects the value at row 0, column 1 (the second column).

```python
# Select the element at row 0, column 1
# Row 0 = first data row (Wireless Mouse)
# Column 1 = second column (Brand)
value = df.iloc[0, 1]

print(f"Value at iloc[0, 1]: {value}")
print(f"Type: {type(value)}")

# display() renders the output nicely in Jupyter
display(value)
```

## 10. Selecting Specific Rows and Columns with `loc[]`

`loc` selects data by **label**. You provide row labels and column names.

**Syntax:** `df.loc[row_labels, column_labels]`

- Row labels can be a single label, a list, or a slice
- Column labels can be a single name, a list, or a slice

Here we select row 0 and three specific columns.

```python
# Select row 0, but only the Name, Brand, and Stock columns
result = df.loc[[0], ['Name', 'Brand', 'Stock']]

print("Selection type:", type(result))
print("Result shape:", result.shape)
display(result)
```

## 11. Boolean Filtering with `loc[]`

You can pass a **boolean Series** (True/False for each row) to `loc` to filter rows. This is one of the most common data selection patterns.

**How it works:**
1. `df['Color'] == 'Black'` creates a boolean mask
2. `df.loc[mask, columns]` returns only rows where the mask is `True`

You can combine multiple conditions with `&` (AND) and `|` (OR).

```python
# Create a boolean mask: True where Color is 'Black'
mask = df['Color'] == 'Black'
print("Boolean mask (first 5):")
print(mask.head())

# Apply the mask with loc to filter rows and select columns
black_products = df.loc[mask, ['Name', 'Brand', 'Stock']]

print(f"\nFound {len(black_products)} black products:")
display(black_products)
```

## 12. Slicing with `iloc[]` — Ranges of Rows and Columns

`iloc` supports Python slice notation: `start:stop` where `stop` is **exclusive**.

`iloc[0:5, 0:3]` means:
- Rows 0, 1, 2, 3, 4 (5 rows total)
- Columns 0, 1, 2 (3 columns total)

```python
# Slice: first 5 rows, first 3 columns
subset = df.iloc[0:5, 0:3]

print(f"Slice shape: {subset.shape}")
print("Columns in slice:", subset.columns.tolist())
display(subset)
```

## 13. Detecting Missing Values with `isnull().sum()`

Missing data is a critical issue in real-world datasets. Pandas provides two paired methods:

- `isnull()` / `isna()` — returns True for each missing value
- `notnull()` / `notna()` — returns True for each present value

Chaining `.sum()` counts the True values per column, giving you a quick missing-data report.

```python
# Check for missing values in each column
missing_report = df.isnull().sum()

print("Missing values per column:")
print(missing_report)

# Total missing values in the entire DataFrame
total_missing = df.isnull().sum().sum()
print(f"\nTotal missing values in dataset: {total_missing}")

# Percentage of missing values per column
missing_pct = (df.isnull().sum() / len(df)) * 100
print("\nMissing value percentage per column:")
print(missing_pct.round(2))
```
