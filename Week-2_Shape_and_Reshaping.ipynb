# Week-2: Shape and Reshaping of NumPy Arrays
Inspecting structure with `ndim`/`shape`/`size`, reshaping, flattening, and transposing.

## a. Dimensions of NumPy Array
`arr.ndim` counts the number of axes.

```python
import numpy as np

arr = np.array([[1, 3, 5, 7], [2, 4, 6, 8], [10, 20, 30, 40]])
print(f'Array:\n{arr}')
print(f'Number of dimensions: {arr.ndim}')
```

## b. Shape of NumPy Array
`arr.shape` returns a tuple of lengths along each axis.

```python
import numpy as np

arr = np.array([[1, 3, 5, 7], [2, 4, 6, 8], [10, 20, 30, 40]])
print(f'Shape: {arr.shape}  (rows, columns)')
```

## c. Size of NumPy Array
`arr.size` is the total element count (product of the shape tuple).

```python
import numpy as np

arr = np.array([[1, 3, 5, 7], [2, 4, 6, 8], [10, 20, 30, 40]])
print(f'Total elements: {arr.size}')
print(f'Verified: {arr.shape[0]} x {arr.shape[1]} = {arr.shape[0] * arr.shape[1]}')
```

## d. Reshaping a NumPy Array
Reshape changes the view of the data without copying; the element count must be conserved.

```python
import numpy as np

nums = np.arange(1, 13)
grid = nums.reshape(3, 4)
print(f'Original (1D, {nums.size} elements):\n{nums}')
print(f'Reshaped to 3x4:\n{grid}')
```

## e. Flattening a NumPy Array
`.flatten()` collapses any array into a 1D copy.

```python
import numpy as np

matrix = np.array([[2, 4, 6], [8, 10, 12], [14, 16, 18]])
print(f'2D array:\n{matrix}')
print(f'Flattened: {matrix.flatten()}')
```

## f. Transpose of a NumPy Array
`.T` flips axes: (rows, cols) -> (cols, rows).

```python
import numpy as np

arr = np.array([[5, 10, 15], [20, 25, 30]])
print(f'Original (shape {arr.shape}):\n{arr}')
print(f'Transposed (shape {arr.T.shape}):\n{arr.T}')
```
