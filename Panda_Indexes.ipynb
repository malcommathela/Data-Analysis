# 📊 Mastering Pandas Indexes

## What You Will Learn
This notebook covers the fundamentals of **DataFrame indexing** in pandas — one of the most powerful features for data manipulation. By the end, you will understand how to:

- Set and reset custom indexes
- Name your index for clarity
- Sort data by its index
- Reorder columns with `reindex`

---

## 1. Import Required Libraries

We only need `pandas` for this tutorial. The conventional alias is `pd`.

```python
import pandas as pd

# Verify the installed version (useful for debugging compatibility issues)
print(f"Pandas version: {pd.__version__}")
```

## 2. Load the Dataset

We read a CSV file containing product information. `pd.read_csv()` automatically assigns a **RangeIndex** (0, 1, 2, ...) to the DataFrame.

> 💡 **Pro Tip:** Always inspect your data immediately after loading to catch formatting issues early.

```python
# Load the products dataset from a CSV file
products = pd.read_csv('products.csv')

# Create a DataFrame explicitly (though read_csv already returns one)
# This step is redundant but shown for educational clarity
df = pd.DataFrame(products)

# Display the full DataFrame
df
```

## 3. Understanding the Default Index

By default, pandas assigns an integer index starting from 0. You can access it via `df.index`. Notice that this index is not part of the data itself — it is merely a **row label**.

```python
# Inspect the current index
print("Current index type:", type(df.index))
print("Index values:", df.index.tolist())
print("\nIndex name:", df.index.name)  # Will be None by default
```

## 4. Setting a Custom Index with `set_index()`

`set_index()` allows you to promote one or more columns to become the DataFrame's index. This is useful when a column contains unique identifiers (like `Brand` in this example, though in real data you'd typically use a unique ID).

| Parameter | Description |
|-----------|-------------|
| `keys` | Column name(s) to use as index |
| `inplace` | If `True`, modifies the DataFrame directly without returning a copy |
| `drop` | If `True` (default), removes the column from the data after setting it as index |

> ⚠️ **Warning:** Setting a non-unique column as index is allowed, but `loc` operations may return multiple rows.

```python
# Set 'Brand' as the index — modifies df in place
df.set_index('Brand', inplace=True)

# Verify the change
print("Index after set_index:", df.index.name)
df.head()
```

## 5. Resetting the Index with `reset_index()`

`reset_index()` converts the current index back into a regular column and restores the default integer index. This is the inverse operation of `set_index()`.

| Parameter | Description |
|-----------|-------------|
| `inplace` | Modify in place or return a copy |
| `drop` | If `True`, discards the index instead of converting it to a column |
| `col_level` | For MultiIndex, determines which level to reset |

```python
# Reset the index — 'Brand' becomes a column again
df.reset_index(inplace=True)

# Verify restoration
print("Index after reset:", df.index.name)
print("Columns now include 'Brand':", 'Brand' in df.columns)
df.head()
```

## 6. Naming Your Index with `index.name`

Giving your index a descriptive name improves readability, especially when exporting to formats like Excel or SQL where the index name becomes a column header.

```python
# Assign a custom name to the index
df.index.name = 'First'

# Display the DataFrame — notice the index label in the top-left corner
print("Index name set to:", df.index.name)
df.head()
```

## 7. Sorting by Index with `sort_index()`

`sort_index()` sorts the DataFrame based on its index values. This is particularly useful after setting a string or datetime index.

| Parameter | Description |
|-----------|-------------|
| `ascending` | `True` = A→Z / 0→9; `False` = Z→A / 9→0 |
| `inplace` | Modify in place or return a copy |
| `kind` | Sorting algorithm: 'quicksort', 'mergesort', 'heapsort', 'stable' |
| `na_position` | Where to place NaN values: 'first' or 'last' |

```python
# Sort the index in descending order (Z → A for strings)
df.sort_index(ascending=False, inplace=True)

# Verify the sort — rows should now appear in reverse index order
print("First few index values after sort:", df.index[:5].tolist())
df.head()
```

## 8. Reordering Columns with `reindex()`

`reindex()` is a versatile method. When applied to **columns**, it returns a new DataFrame with only the specified columns, in the specified order. Columns not listed are dropped.

> 📝 **Note:** The original DataFrame `df` is NOT modified unless you reassign.

| Parameter | Description |
|-----------|-------------|
| `columns` | List of column names in desired order |
| `fill_value` | Value to use for missing data when rows/columns are added |
| `method` | Interpolation method for filling missing values |

```python
# Reorder columns — only keep 'last', 'first', 'email'
# Note: These column names don't exist in our dataset, so this will error.
# Let's use actual column names instead for a working example:

# First, let's see what columns we actually have
print("Available columns:", df.columns.tolist())

# Reindex to select and reorder a subset of columns
df_reordered = df.reindex(columns=['Stock', 'Name', 'Brand'])

# Display the result
df_reordered.head()
```

---

## 🎯 Summary

| Method | Purpose | In-Place? |
|--------|---------|-----------|
| `set_index()` | Promote column(s) to index | Yes (with `inplace=True`) |
| `reset_index()` | Restore default integer index | Yes (with `inplace=True`) |
| `index.name` | Assign a descriptive label to the index | Yes (direct assignment) |
| `sort_index()` | Sort rows by index values | Yes (with `inplace=True`) |
| `reindex(columns=...)` | Reorder/select columns | No (returns copy) |
