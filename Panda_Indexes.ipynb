```python
import pandas as pd
```

```python
products = pd.read_csv('products.csv')
df = pd.DataFrame(products)
```

```python
df
```

```python
df.set_index('Brand', inplace=True)
```

```python
df.reset_index(inplace=True)
```

```python
df.index.name = 'First'
```

```python
df.sort_index(ascending=False, inplace=True)
```
