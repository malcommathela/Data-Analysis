# **Axes**

```python
import numpy as np

## 1d Array
arr_1d = np.array([1,2,3,4])
print("1D Array:\n", arr_1d)
print("Sum (Axis 0): ", np.sum(arr_1d, axis = 0))

## 2D
arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D Array:\n", arr_2d)
print("Sum (Axis 0): ", np.sum(arr_2d, axis = 0))
print("Sum (Axis 1): ", np.sum(arr_2d, axis = 1))

#3D
arr_3d = np.array([[[1, 2], [3, 4]],
                   [[5, 6], [7, 8]]])
print("\n3D Array:\n", arr_3d)
print("Sum along layers (Axis 0): \n", np.sum(arr_3d, axis = 0))
print("Sum along Rows (Axis 0): \n", np.sum(arr_3d, axis = 1))
print("Sum along Columns (Axis 0): \n", np.sum(arr_3d, axis = 2))
print("Overall Sum: ", np.sum(arr_3d))




```

# **Expanding a NumPy array**

Syntax

```python
np.expand_dims(array, axis)
```

Example 1: Convert 1D array to Row Vector

```python
import numpy as np

arr = np.array([1, 2, 3, 4])

expanded = np.expand_dims(arr, axis=0)

print(expanded)
print(expanded.shape)
```

Example 2: Convert 1D array to Column Vector

```python
import numpy as np

arr = np.array([1, 2, 3, 4])

expanded = np.expand_dims(arr, axis=1)

print(expanded)
print(expanded.shape)
```

Example 3: Expanding a 2D Array to 3D

```python
import numpy as np

arr = np.array([[1, 2],
                [3, 4]])

expanded = np.expand_dims(arr, axis=0)

print(expanded)
print(expanded.shape)
```

# Squeezing a NumPy **array**

Syntax

```python
np.squeeze(array, axis=None)
```

Example 1: Squeezing a 2D Array

```python
import numpy as np

arr = np.array([[1, 2, 3]])

print("Original Shape:", arr.shape)

squeezed = np.squeeze(arr)

print("Squeezed Array:")
print(squeezed)
print("New Shape:", squeezed.shape)
```

Example 2: Squeezing a 3D Array

```python
import numpy as np

arr = np.array([[[1, 2],
                 [3, 4]]])

print("Original Shape:", arr.shape)

squeezed = np.squeeze(arr)

print(squeezed)
print("New Shape:", squeezed.shape)
```

Example 3: Using axis

```python
import numpy as np

arr = np.array([[[10, 20, 30]]])

print(arr.shape)

result = np.squeeze(arr, axis=0)

print(result)
print(result.shape)
```

# Sorting in NumPy **Arrays**

Syntax

```python
np.sort(array, axis=-1)
```

Sorting a 1D Array

```python
import numpy as np

arr = np.array([5, 2, 8, 1, 9])

sorted_arr = np.sort(arr)

print("Original Array:", arr)
print("Sorted Array:", sorted_arr)
```

Sorting a 2D Array (Rows)

```python
import numpy as np

arr = np.array([[3, 1, 2],
                [9, 5, 7]])

print(np.sort(arr, axis=1))
```

Sorting a 2D Array (Columns)

```python
import numpy as np

arr = np.array([[3, 1, 2],
                [9, 5, 7]])

print(np.sort(arr, axis=0))
```

Sorting a 3D Array

```python
import numpy as np

arr = np.array([
    [[3, 1], [4, 2]],
    [[8, 6], [7, 5]]
])

print(np.sort(arr, axis=2))
```
