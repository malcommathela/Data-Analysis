# Week-9: Reading Pickle, Image, Multiple Files and Database

Binary serialization, image loading, batch file reading and SQL extraction.

## a. Read Pickle Files

Pickle restores the exact DataFrame object, including data types.

```python
import pandas as pd

pkl_path = "datasets/student.pkl"

students_pkl = pd.read_pickle(pkl_path)
students_pkl
```

## b. Read Image Files using PIL

We open the image, inspect its properties, and display it inline.

```python
from PIL import Image

image_path = "datasets/flower.jpg"

flower = Image.open(image_path)
print(f"Format: {flower.format} | Size: {flower.size} | Mode: {flower.mode}")
display(flower)
```

## c. Read Multiple Files using Glob

A wildcard pattern collects every matching CSV in the folder; we then load and preview each one.

```python
import glob

csv_paths = sorted(glob.glob("datasets/records_*.csv"))

for path in csv_paths:
    frame = pd.read_csv(path)
    print(f"--- {path} ---")
    display(frame.head(2))
```

## d. Import Data from Database

We connect to SQLite, pull the `Student` table with a SQL query, and close the connection automatically with a context manager.

```python
import sqlite3

db_path = "datasets/student.db"

with sqlite3.connect(db_path) as connection:
    students_db = pd.read_sql_query("SELECT RollNo, Name, Marks FROM Student", connection)

students_db
```
