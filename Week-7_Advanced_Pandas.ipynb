# Week-7: Advanced Pandas Operations
Missing-value handling, sorting by column values, and split-apply-combine aggregation with `groupby`.

## a. Filling NaN with String
Missing values (`np.nan`) are located with `.isna()` and replaced with `.fillna()`.

```python
import pandas as pd
import numpy as np

data = {
    'Student': ['Asha', 'Vikram', np.nan, 'Meena', 'Kiran'],
    'Branch':  ['CSE', np.nan, 'IT', 'AIML', np.nan]
}
df = pd.DataFrame(data)
print('Missing values per column:')
print(df.isna().sum())

print('\nAfter fillna("Not Available"):')
print(df.fillna('Not Available'))
```

## b. Sorting Based on Column Values
`sort_values` returns a new sorted DataFrame; `ascending=False` reverses the order.

```python
import pandas as pd

df = pd.DataFrame({
    'Student': ['Asha', 'Vikram', 'Meena', 'Kiran'],
    'Score':   [72, 94, 81, 66]
})
print('Original:')
print(df)

print('\nAscending by Score:')
print(df.sort_values(by='Score'))

print('\nDescending by Score:')
print(df.sort_values(by='Score', ascending=False))
```

## c. groupby()
Group rows by a key column, then aggregate a metric column (split-apply-combine).

```python
import pandas as pd

df = pd.DataFrame({
    'Branch': ['IT', 'CSE', 'IT', 'AIML', 'CSE', 'AIML'],
    'Score':  [70, 88, 76, 92, 84, 95]
})
print('Dataset:')
print(df)

print('\nAverage score per branch:')
print(df.groupby('Branch')['Score'].mean())

print('\nTotal score per branch:')
print(df.groupby('Branch')['Score'].sum())

print('\nCount of students per branch:')
print(df.groupby('Branch')['Score'].count())
```
