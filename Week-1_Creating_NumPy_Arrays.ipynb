# Week-1: Creating NumPy Arrays
Array creation with `np.array`, `zeros`, `ones`, random generators, `identity`, and `linspace`.

## a. Create a Basic ndarray
Convert a Python list into a 1D ndarray with `np.array()`.

```python
import numpy as np

values = [5, 15, 25, 35, 45]
arr = np.array(values)
print(f'Array: {arr} | dtype: {arr.dtype} | shape: {arr.shape}')
```

## b. Create an Array of Zeros
Pre-allocate a 2D buffer with `np.zeros()` (float64 by default).

```python
import numpy as np

zeros_arr = np.zeros((4, 3))
print(zeros_arr)
```

## c. Create an Array of Ones
`np.ones()` fills any shape with 1.0; `dtype` can be overridden.

```python
import numpy as np

ones_arr = np.ones((3, 2), dtype=np.int32)
print(ones_arr)
print(f'dtype: {ones_arr.dtype}')
```

## d. Generate Random Numbers in ndarray
Use the modern recommended generator `np.random.default_rng()` with a seed for reproducibility.

```python
import numpy as np

rng = np.random.default_rng(seed=42)
rand_arr = rng.integers(10, 99, size=(3, 3))
print(rand_arr)
```

## e. Create an Array of Your Choice
A manually specified 1D array (squares of the first even numbers).

```python
import numpy as np

my_arr = np.array([4, 16, 36, 64, 100])
print(my_arr)
```

## f. Create an Identity Matrix
`np.identity(n)` gives an n×n matrix with 1.0 on the main diagonal.

```python
import numpy as np

identity_mat = np.identity(5)
print(identity_mat)
```

## g. Create an Evenly Spaced ndarray
`np.linspace` guarantees an exact number of points **including** the stop value.

```python
import numpy as np

points = np.linspace(2, 14, 7)
print(points)
print(f'Step size: {points[1] - points[0]}')
```
