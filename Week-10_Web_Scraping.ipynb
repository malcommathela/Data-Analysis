# Week-10: Web Scraping using Python

Sending HTTP requests, checking status codes, parsing HTML with BeautifulSoup and extracting elements.

> **Note:** These cells require an internet connection to run.

## a. Sending a Request to a Website

```python
import requests

url = "https://quotes.toscrape.com"
response = requests.get(url, timeout=10)

if response.status_code == 200:
    print("Request successful. Status:", response.status_code)
else:
    print("Request failed. Status:", response.status_code)
```

## b. Parsing HTML using BeautifulSoup

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")
print("Page Title:", soup.title.get_text(strip=True))
```

## c. Extracting Headings and Paragraphs

```python
heading = soup.find("h1")
paragraph = soup.find("p")

print("Heading :", heading.get_text(strip=True) if heading else "No heading found")
print("Paragraph:", paragraph.get_text(strip=True) if paragraph else "No paragraph found")
```

## d. Extracting Multiple Elements

Scraping every book title listed on the demo bookstore site.

```python
store_url = "https://books.toscrape.com/"
store_response = requests.get(store_url, timeout=10)

store_soup = BeautifulSoup(store_response.text, "html.parser")
book_titles = store_soup.find_all("h3")

print(f"Found {len(book_titles)} books:\n")
for i, book in enumerate(book_titles, start=1):
    print(i, "-", book.get_text(strip=True))
```
