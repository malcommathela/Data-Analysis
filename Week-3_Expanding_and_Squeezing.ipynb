# Week-3: Expanding and Squeezing NumPy Arrays
Adding/removing singleton dimensions and sorting along axes.

## a. Expanding a NumPy Array
`np.expand_dims` inserts a length-1 axis; the same can be written with `np.newaxis` indexing.

```python
import numpy as np

arr = np.array([5, 10, 15, 20])
print(f'Original shape: {arr.shape}')

row_vec = np.expand_dims(arr, axis=0)
print(f'After expand_dims(axis=0) -> shape {row_vec.shape}:\n{row_vec}')

col_vec = arr[:, np.newaxis]
print(f'After [:, np.newaxis] -> shape {col_vec.shape}:\n{col_vec}')
```

## b. Squeezing a NumPy Array
`np.squeeze` removes all length-1 axes.

```python
import numpy as np

arr = np.array([[[[3, 6, 9, 12]]]])
print(f'Original shape: {arr.shape}')

squeezed = np.squeeze(arr)
print(f'Squeezed shape: {squeezed.shape}')
print(f'Squeezed array: {squeezed}')
```

## c. Sorting in NumPy Arrays
`np.sort` returns a sorted copy (ascending by default); slice `[::-1]` for descending.

```python
import numpy as np

arr1d = np.array([58, 13, 91, 27, 44, 6])
print(f'Original 1D: {arr1d}')
print(f'Sorted ascending:  {np.sort(arr1d)}')
print(f'Sorted descending: {np.sort(arr1d)[::-1]}')

arr2d = np.array([[58, 13, 91], [27, 44, 6]])
print(f'Original 2D:\n{arr2d}')
print(f'Sorted along rows (axis=1):\n{np.sort(arr2d, axis=1)}')
print(f'Sorted down columns (axis=0):\n{np.sort(arr2d, axis=0)}')
```
