# Week-11: Data Preprocessing Techniques

Scaling, standardization and encoding features from the loan prediction dataset.

## a. Feature Scaling (MinMaxScaler)

Rescales `ApplicantIncome` into the range [0, 1].

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

loans = pd.read_csv("datasets/loan_prediction.csv")

minmax = MinMaxScaler()
loans["Income_MinMax"] = minmax.fit_transform(loans[["ApplicantIncome"]])

loans[["ApplicantIncome", "Income_MinMax"]].head()
```

## b. Feature Standardization (StandardScaler)

Transforms `LoanAmount` to zero mean and unit variance.

```python
from sklearn.preprocessing import StandardScaler

standard = StandardScaler()
loans["Loan_Standardized"] = standard.fit_transform(loans[["LoanAmount"]])

loans[["LoanAmount", "Loan_Standardized"]].head()
```

## c. Label Encoding

Converts the `Gender` strings into integer labels (0/1).

```python
from sklearn.preprocessing import LabelEncoder

gender_encoder = LabelEncoder()
loans["Gender_Label"] = gender_encoder.fit_transform(loans["Gender"])

print("Classes:", list(gender_encoder.classes_))
loans[["Gender", "Gender_Label"]].head()
```

## d. One Hot Encoding

Splits the nominal `Property_Area` column into binary indicator columns (`dtype=int` keeps them as 0/1 integers).

```python
loans_encoded = pd.get_dummies(loans, columns=["Property_Area"], dtype=int)

loans_encoded.head()
```
