# Week-12: Data Visualization using Matplotlib

Bar, pie, box, histogram, line/subplots and scatter charts.

## a. Bar Graph

```python
import matplotlib.pyplot as plt

courses = ["Python", "Java", "C", "C++"]
enrolled = [60, 45, 35, 50]

plt.figure(figsize=(6, 4))
plt.bar(courses, enrolled, color="teal", edgecolor="black")
plt.title("Students Enrolled per Course")
plt.xlabel("Course")
plt.ylabel("Number of Students")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

## b. Pie Chart

```python
languages = ["Python", "Java", "C++", "C"]
share = [45, 25, 20, 10]

plt.figure(figsize=(5, 5))
plt.pie(share, labels=languages, autopct="%1.0f%%", startangle=90,
        colors=["#4CAF50", "#2196F3", "#FF9800", "#9E9E9E"])
plt.title("Programming Language Preference")
plt.show()
```

## c. Box Plot

```python
scores = [65, 70, 72, 75, 80, 82, 85, 90, 95, 40, 38]

plt.figure(figsize=(5, 4))
plt.boxplot(scores, tick_labels=["Class A"], showmeans=True)
plt.title("Test Score Distribution")
plt.ylabel("Marks")
plt.grid(axis="y", linestyle=":", alpha=0.5)
plt.show()
```

## d. Histogram

```python
plt.figure(figsize=(6, 4))
plt.hist(scores, bins=5, color="orange", edgecolor="black")
plt.title("Frequency of Marks")
plt.xlabel("Marks Range")
plt.ylabel("Frequency")
plt.show()
```

## e. Line Chart and Subplots

```python
weeks = [1, 2, 3, 4, 5]
revenue = [120, 150, 180, 170, 210]
expenses = [80, 95, 110, 105, 130]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(weeks, revenue, marker="o", color="green", label="Revenue")
plt.plot(weeks, expenses, marker="s", color="red", label="Expenses")
plt.title("Weekly Finances")
plt.xlabel("Week")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(weeks, [r - e for r, e in zip(revenue, expenses)],
         marker="^", color="purple")
plt.title("Net Profit per Week")
plt.xlabel("Week")

plt.tight_layout()
plt.show()
```

## f. Scatter Plot

```python
heights = [150, 155, 160, 165, 170, 175]
weights = [45, 50, 55, 60, 65, 70]

plt.figure(figsize=(6, 4))
plt.scatter(heights, weights, color="crimson", marker="D")
plt.title("Height vs Weight Correlation")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
